```python
#program 14
n = int(input())

last = n % 10

temp = n
while temp >= 10:
    temp = temp // 10

first = temp

digits = 0
temp = n

while temp > 0:
    digits = digits + 1
    temp = temp // 10

power = 10 ** (digits - 1)

middle = n % power
middle = middle // 10

result = last * power + middle * 10 + first

print(result)
```

```python
#program 17
n = int(input())

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count = count + 1

sum = 0
temp = n

while temp > 0:
    sum = sum + temp % 10
    temp = temp // 10

if count == 2 and sum == 14:
    print("Prime & Sum of Digits is 14")
elif count != 2 and sum == 14:
    print("Not Prime but sum of digits is 14")
elif count == 2:
    print("Prime, but sum of Digits is not 14")
else:
    print("Not Prime and sum of Digits is not 14")
```

```python
#program 28
a, b = map(int, input().split())

if a > b:
    lcm = a
else:
    lcm = b

while True:
    if lcm % a == 0 and lcm % b == 0:
        print(lcm)
        break

    lcm = lcm + 1
```

```python
#program 29
a, b, c = map(int, input().split())

if a > b and a > c:
    lcm = a
elif b > c:
    lcm = b
else:
    lcm = c

while True:
    if lcm % a == 0 and lcm % b == 0 and lcm % c == 0:
        print(lcm)
        break

    lcm = lcm + 1
```
