```python
#program 3
def check_sum(n):
    sum = 0

    while n > 0:
        sum = sum + n % 10
        n = n // 10

    if sum == 14:
        print("Sum of Digits is 14")
    else:
        print("Sum of Digits is not 14")

n = int(input())
check_sum(n)
```

```python
#program 4
def check_prime(n):
    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1

    if count == 2:
        print("Number is Prime")
    else:
        print("Number is not Prime")

n = int(input())
check_prime(n)
```

```python
#program 5
n = int(input())
count = 0

while n > 0:
    if n % 10 == 0:
        count = count + 1

    n = n // 10

print(count)
```
