```python
#program 32
count = 0

for i in range(2, 1000000):
    prime = True

    for j in range(2, int(i ** 0.5) + 1):
        if i % j == 0:
            prime = False
            break

    if prime:
        n = i
        sum = 0

        while n > 0:
            sum = sum + n % 10
            n = n // 10

        if sum == 14:
            count = count + 1

print(count)
```

```python
#program 48
arr = list(map(int, input().split()))

for i in range(len(arr) - 1, 0, -1):
    if arr[i] >= 10:
        carry = arr[i] // 10
        arr[i] = arr[i] % 10
        arr[i - 1] = arr[i - 1] + carry

print(*arr)
```
